```python
import jax
import jax.numpy as jnp
import jax.scipy as jsp
import numpy as np
from magi import MAGISolver
```

```python
import warnings
warnings.filterwarnings("error")
```

```python
def ode(X, theta, t=None):
    V, R = X
    a, b, c = theta
    return jnp.stack([c * (V - V**3/3 + R), -1/c * (V - a + b*R)])
```

```python
def discretize_data(sample, I, rounded=3):
    '''
    Discretize a dataset at time set I. Rounds times in I to `rounded` decimal places to mitigate floating point errors.
    '''
    I = np.round(I, rounded)
    data_disc = np.full(shape=(len(I), sample.shape[1]), fill_value=np.nan)
    data_disc[:,0] = I
    data_disc[np.isin(data_disc[:,0], sample[:,0])] = sample
    return data_disc
```

```python
y = np.loadtxt('y.csv', delimiter=',')
data = discretize_data(y, np.arange(0, 20.1, 0.25))
# data[:,2] = np.nan
```

```python
solver = MAGISolver(ode, data, jnp.ones(3))
```
